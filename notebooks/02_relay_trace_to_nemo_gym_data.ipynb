# Relay Trace To NeMo Gym Data

This notebook converts Relay traces into NeMo Gym JSONL rows with `responses_create_params` and `task_data`, then stages the generated files for the local NeMo Gym resources server.

```python
from pathlib import Path
from shutil import copyfile

ROOT = Path.cwd()
GYM_OUTPUT = ROOT / "outputs" / "nemo_gym"
SERVER_DATA = ROOT / "resources_servers" / "relay_handoff" / "data"
GYM_OUTPUT.mkdir(parents=True, exist_ok=True)
SERVER_DATA.mkdir(parents=True, exist_ok=True)
```

Create NeMo Gym data from the latest managed Relay trace.

Copy train, validation, and example JSONL into the resources server data directory used by the NeMo Gym config.

```python
for name in ["example.jsonl", "train.jsonl", "validation.jsonl"]:
    copyfile(GYM_OUTPUT / name, SERVER_DATA / name)
    print(SERVER_DATA / name)
```
